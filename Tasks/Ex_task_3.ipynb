### **Пункт №2. Обработка исключений в цикле**

Напишите программу, которая запрашивает у пользователя числа и добавляет их в список:

- при некорректном вводе программа выводит сообщение об ошибке и продолжает работу;
- для выхода пользователь вводит слово «стоп».

**Обязательно:**

- покажите ввод минимум 7 значений, включая ошибочные;
- выведите итоговый список и статистику: сколько чисел введено успешно, сколько было ошибок.

Пример работы:

```
Введите число: 10
Введите число: текст
Ошибка: это не число
Введите число: стоп
Итоговый список: [10]
```

````python
numbers = []
ok = 0
errors = 0

while True:
    value = input('Введите число: ')
    if value == 'стоп':
        break
    try:
        number = int(value)
        numbers.append(number)
        ok = ok + 1
    except ValueError:
        print('Ошибка: это не число')
        errors = errors + 1

print('Итоговый список:', numbers)
print('Успешно введено:', ok)
print('Ошибок:', errors)

````

### **Пункт №3. Безопасное получение элемента списка**

Создайте функцию `get_list_element(lst, index)`, которая безопасно извлекает элемент списка по индексу. Используйте все четыре блока: `try`, `except`, `else` и `finally`.

Функция должна:

- обрабатывать выход за пределы списка;
- обрабатывать индекс неподходящего типа;
- в блоке `finally` выводить сообщение «Попытка доступа завершена».

**Обязательно:**

- выведите результаты минимум для 4 случаев: успешный доступ, выход за границы, отрицательный индекс, нечисловой индекс;
- покажите, что блок `finally` выполняется всегда.

Пример использования:

```python
my_list = [1, 2, 3, 4, 5]
get_list_element(my_list, 2)   # Элемент: 3 / Попытка доступа завершена
get_list_element(my_list, 10)  # Индекс вне диапазона / Попытка доступа завершена
```

````python
def get_list_element(lst, index):
    try:
        result = lst[index]
    except IndexError:
        print('Индекс вне диапазона')
    except TypeError:
        print('Индекс неподходящего типа')
    else:
        print('Элемент:', result)
    finally:
        print('Попытка доступа завершена')


my_list = [1, 2, 3, 4, 5]
print('Список:', my_list)
for index in [2, 10, -1, 'текст']:
    print('Индекс:', index)
    get_list_element(my_list, index)
    print()

````

````python
with open('output.txt', 'w', encoding='utf-8') as file:
    file.write('тюлень спит тюлень отдыхает\nвкусный чизбургер на обед\nтут картошка фри\nстрельцов пошел гулять\nоткрой гитхаб и сделай коммит\nмама варит вкусный суп\nтюлень любит рыбу\nкартошка очень вкусная\nмама пришла домой\nэто конец файла\n')

print('Содержимое файла:')
with open('output.txt', 'r', encoding='utf-8') as file:
    print(file.read())

words = ['тюлень', 'чизбургер', 'картошка', 'стрельцов', 'гитхаб', 'мама']
words.append(input('Введите слово для поиска: ').strip())

for word in words:
    print('Ищем слово:', word)
    count = 0
    with open('output.txt', 'r', encoding='utf-8') as file:
        for number, line in enumerate(file, start=1):
            line_count = 0
            for item in line.split():
                if item.strip('.,!?;:—-()[]{}«»\"\'').lower() == word.lower():
                    line_count = line_count + 1
            if line_count > 0:
                print('Найдено в строке:', number)
                count = count + line_count
    print('Всего вхождений:', count)
    print()

````

### **Пункт №7. Подсчёт статистики файла**

Напишите функцию `file_statistics(filename)`, которая читает текстовый файл и возвращает словарь со статистикой:

- количество строк;
- количество слов;
- количество символов с пробелами;
- количество символов без пробелов (пробелы и переводы строк не учитываются).

**Обязательно:**

- создайте тестовый файл минимум с 5 строками текста;
- выведите содержимое файла до обработки;
- выведите полученную статистику;
- обработайте случай несуществующего файла.

Текст для тестового файла:

```python
test_content = """Первая строка текста
Вторая строка с большим количеством слов
Третья строка
Четвертая строка для теста
Пятая и последняя строка"""
```

````python
def file_statistics(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            lines = file.readlines()
    except FileNotFoundError:
        print('Файла нет!')
        return None

    text = ''.join(lines)
    chars_no_spaces = 0
    for char in text:
        if char != ' ' and char != '\n':
            chars_no_spaces = chars_no_spaces + 1

    return {
        'строк': len(lines),
        'слов': len(text.split()),
        'символов': len(text),
        'без пробелов': chars_no_spaces
    }


with open('test_stats.txt', 'w', encoding='utf-8') as file:
    file.write('Первая строка текста\nВторая строка с большим количеством слов\nТретья строка\nЧетвертая строка для теста\nПятая и последняя строка\n')

print('Что записано в файле:')
with open('test_stats.txt', 'r', encoding='utf-8') as file:
    print(file.read())
print('Статистика:', file_statistics('test_stats.txt'))

with open('line_test.txt', 'w', encoding='utf-8') as file:
    file.write('a\n\nb\n')
print('Проверяем текст:', repr('a\n\nb\n'))
print('Статистика:', file_statistics('line_test.txt'))

print('Проверяем отсутствующий файл:')
print('Статистика:', file_statistics('missing_stats_example.txt'))

````

### **Пункт №8. Поиск и замена в файле**

Напишите функцию `replace_in_file(filename, old_text, new_text)`, которая заменяет все вхождения `old_text` на `new_text` и сохраняет результат в новый файл с префиксом `modified_` (например, `modified_output.txt`).

**Обязательно:**

- создайте файл с текстом, в котором заменяемое слово встречается минимум 3 раза;
- выведите содержимое файла до замены;
- выполните замену;
- выведите содержимое нового файла после замены;
- выведите количество произведённых замен.

````python
def replace_in_file(filename, old_text, new_text):
    with open(filename, 'r', encoding='utf-8') as file:
        content = file.read()

    count = content.count(old_text)
    modified_content = content.replace(old_text, new_text)

    with open('modified_' + filename, 'w', encoding='utf-8') as file:
        file.write(modified_content)
    return count


with open('original.txt', 'w', encoding='utf-8') as file:
    file.write('привет мир\nтут есть слово тест один раз\nеще раз слово тест тут\nи третий раз слово тест в конце\nпросто строка\n')

print('Файл до замены:')
with open('original.txt', 'r', encoding='utf-8') as file:
    print(file.read())

print('Замен сделано:', replace_in_file('original.txt', 'тест', 'экзамен'))
print('Новый файл:')
with open('modified_original.txt', 'r', encoding='utf-8') as file:
    print(file.read())

````

### **Пункт №9. Объединение файлов**

Создайте функцию `merge_files(file_list, output_file)`, которая объединяет содержимое нескольких текстовых файлов в один и добавляет после содержимого каждого файла разделитель.

**Обязательно:**

- создайте минимум 3 файла с разным содержимым;
- выведите содержимое каждого исходного файла;
- выполните объединение;
- выведите содержимое итогового файла.

````python
def merge_files(file_list, output_file):
    with open(output_file, 'w', encoding='utf-8') as output:
        for name in file_list:
            with open(name, 'r', encoding='utf-8') as file:
                text = file.read()
            output.write(text)
            if text and not text.endswith('\n'):
                output.write('\n')
            output.write('--- РАЗДЕЛИТЕЛЬ ---\n')


with open('file1.txt', 'w', encoding='utf-8') as file:
    file.write('Это текст из первого файла\n')
with open('file2.txt', 'w', encoding='utf-8') as file:
    file.write('Тут информация для второго файла\n')
with open('file3.txt', 'w', encoding='utf-8') as file:
    file.write('А здесь концовка в третьем файле\n')

file_list = ['file1.txt', 'file2.txt', 'file3.txt']
for name in file_list:
    print('Содержимое', name)
    with open(name, 'r', encoding='utf-8') as file:
        print(file.read())

merge_files(file_list, 'merged_output.txt')
print('Содержимое объединенного файла:')
with open('merged_output.txt', 'r', encoding='utf-8') as file:
    print(file.read())
````

````python
# @title Результаты автоматической проверки
# AUTOCHECKER_A3
import contextlib as _contextlib
import html as _html
import io as _io
import json as _json
import sys as _sys
import types as _types
import urllib.request as _urllib

from IPython.display import HTML as _HTML, display as _display

_TASK_IDS = ['a3_01',
 'a3_02',
 'a3_03',
 'a3_04',
 'a3_05',
 'a3_06',
 'a3_07',
 'a3_08',
 'a3_09',
 'a3_10',
 'a3_11',
 'a3_12']
_TASK_SOURCE_IDS = {'a3_01': ['724c5954'],
 'a3_02': ['31448e09'],
 'a3_03': ['1c067cf3'],
 'a3_04': ['cbce6c5f'],
 'a3_05': ['df27a986'],
 'a3_06': ['84aa205a'],
 'a3_07': ['31bd8a06'],
 'a3_08': ['d961bbf3'],
 'a3_09': ['0c262bd2'],
 'a3_10': ['43bfd6b8'],
 'a3_11': ['64a05e35'],
 'a3_12': ['581b5868']}
_TASK_EDITABLE_IDS = {'a3_01': ['724c5954'],
 'a3_02': ['31448e09'],
 'a3_03': ['1c067cf3'],
 'a3_04': ['cbce6c5f'],
 'a3_05': ['df27a986'],
 'a3_06': ['84aa205a'],
 'a3_07': ['31bd8a06'],
 'a3_08': ['d961bbf3'],
 'a3_09': ['0c262bd2'],
 'a3_10': ['43bfd6b8'],
 'a3_11': ['64a05e35'],
 'a3_12': ['581b5868']}
_TASK_SOURCE_POSITIONS = {'a3_01': [0],
 'a3_02': [1],
 'a3_03': [2],
 'a3_04': [3],
 'a3_05': [4],
 'a3_06': [5],
 'a3_07': [6],
 'a3_08': [7],
 'a3_09': [8],
 'a3_10': [9],
 'a3_11': [10],
 'a3_12': [11]}
_TASK_EDITABLE_POSITIONS = {'a3_01': [0],
 'a3_02': [1],
 'a3_03': [2],
 'a3_04': [3],
 'a3_05': [4],
 'a3_06': [5],
 'a3_07': [6],
 'a3_08': [7],
 'a3_09': [8],
 'a3_10': [9],
 'a3_11': [10],
 'a3_12': [11]}
_CHECKER_CELL_ID = "autoCheckA3"
_CHECKER_URL = "https://raw.githubusercontent.com/tester170/python_programming_autotest/main/answer_checker.py"


def _notebook_from_reply(value):
    if isinstance(value, str):
        value = _json.loads(value)
    if not isinstance(value, dict):
        raise TypeError("Colab вернул данные блокнота в неизвестном формате")
    if isinstance(value.get("cells"), list):
        return value
    for key in ("ipynb", "notebook", "data"):
        nested = value.get(key)
        if isinstance(nested, str):
            try:
                nested = _json.loads(nested)
            except _json.JSONDecodeError:
                continue
        if isinstance(nested, dict) and isinstance(nested.get("cells"), list):
            return nested
    raise ValueError("Не удалось получить ячейки открытого блокнота")


def _cell_source(cell):
    source = cell.get("source", "")
    return "".join(source) if isinstance(source, list) else str(source)


def _cell_id(cell):
    return cell.get("id") or cell.get("metadata", {}).get("id")


def _solution_is_empty(code):
    return not any(
        line.strip() and not line.lstrip().startswith("#")
        for line in code.splitlines()
    )


def _status(result):
    if result is None:
        return "⚠️ Решение не заполнено"
    if result["ok"]:
        return f"✅ Верно ({result['passed']}/{result['total']})"
    if result["error"] == "timeout":
        return f"⏱️ Превышено время выполнения ({result['passed']}/{result['total']})"
    if result["error"] == "runtime":
        return f"❌ Ошибка выполнения ({result['passed']}/{result['total']})"
    return f"❌ Есть ошибка ({result['passed']}/{result['total']})"


try:
    from google.colab import _message as _colab_message

    _reply = _colab_message.blocking_request("get_ipynb", request="", timeout_sec=30)
    _notebook = _notebook_from_reply(_reply)

    _request = _urllib.Request(
        _CHECKER_URL,
        headers={"User-Agent": "python-programming-autochecker"},
    )
    with _urllib.urlopen(_request, timeout=20) as _response:
        _checker_source = _response.read().decode("utf-8")

    _checker_name = "_python_programming_autochecker"
    _checker = _types.ModuleType(_checker_name)
    _checker.__file__ = _CHECKER_URL
    _sys.modules[_checker_name] = _checker
    with _contextlib.redirect_stdout(_io.StringIO()), _contextlib.redirect_stderr(_io.StringIO()):
        exec(compile(_checker_source, _CHECKER_URL, "exec"), _checker.__dict__)

    if getattr(_checker, "CHECKER_VERSION", 0) < 3:
        raise RuntimeError("Сначала замените содержимое репозитория автопроверки файлами из подготовленного архива")
    _missing_tasks = [task_id for task_id in _TASK_IDS if task_id not in _checker.TASKS]
    if _missing_tasks:
        raise RuntimeError("В репозитории автопроверки отсутствуют тесты для этого блокнота")

    _code_cells = []
    _code_by_id = {}
    for _cell in _notebook.get("cells", []):
        if _cell.get("cell_type") != "code":
            continue
        _source = _cell_source(_cell)
        _id = _cell_id(_cell)
        if _id == _CHECKER_CELL_ID or "AUTOCHECKER_" in _source:
            continue
        _code_cells.append(_source)
        if _id:
            _code_by_id[_id] = _source

    def _resolve_sources(ids, positions):
        by_id = [_code_by_id.get(cell_id) for cell_id in ids]
        if all(source is not None for source in by_id):
            return by_id
        try:
            return [_code_cells[position] for position in positions]
        except IndexError as error:
            raise RuntimeError("Не удалось найти ячейки заданий в блокноте") from error

    _rows = []
    _passed_tasks = 0
    for _index, _task_id in enumerate(_TASK_IDS, 1):
        _sources = _resolve_sources(
            _TASK_SOURCE_IDS[_task_id], _TASK_SOURCE_POSITIONS[_task_id]
        )
        _editable = _resolve_sources(
            _TASK_EDITABLE_IDS[_task_id], _TASK_EDITABLE_POSITIONS[_task_id]
        )
        _code = "\n\n".join(_sources)
        _result = None if _solution_is_empty("\n\n".join(_editable)) else _checker.check_task(_task_id, _code)
        if _result is not None and _result["ok"]:
            _passed_tasks += 1
        _title = _checker.TASKS[_task_id].title
        _rows.append(
            "<tr>"
            f"<td style='padding:7px 10px;border-bottom:1px solid #eee'>{_index}</td>"
            f"<td style='padding:7px 10px;border-bottom:1px solid #eee'>{_html.escape(_title)}</td>"
            f"<td style='padding:7px 10px;border-bottom:1px solid #eee'>{_html.escape(_status(_result))}</td>"
            "</tr>"
        )

    _report = f"""
    <div style="font-family:Arial,sans-serif;max-width:850px">
      <h3 style="margin:0 0 8px">Результаты автоматической проверки</h3>
      <table style="border-collapse:collapse;width:100%;border:1px solid #ddd">
        <thead><tr style="background:#f5f5f5">
          <th style="padding:8px;text-align:left">№</th>
          <th style="padding:8px;text-align:left">Задание</th>
          <th style="padding:8px;text-align:left">Результат</th>
        </tr></thead>
        <tbody>{''.join(_rows)}</tbody>
      </table>
      <div style="margin-top:12px;font-size:18px"><b>Итого: {_passed_tasks} из {len(_TASK_IDS)}</b></div>
    </div>
    """
    _display(_HTML(_report))
except Exception as _error:
    _display(_HTML(
        "<div style='font-family:Arial,sans-serif;padding:12px;border:1px solid #e0a0a0;background:#fff7f7'>"
        "⚠️ Не удалось запустить автоматическую проверку: "
        + _html.escape(str(_error))
        + "</div>"
    ))
````
